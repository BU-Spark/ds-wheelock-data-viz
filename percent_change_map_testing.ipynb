```python
def aggregate_pre_pandemic(pre_pandemic_bullying_type_data):
    months = ['January', "February", "March", "April", "May", "June", "July", "August", "September", "October",
             "November", "December"]

    aggregate = pd.DataFrame()

    for idx, month in enumerate(months):
        """Get all cols for x month and average"""
        subset = pre_pandemic_bullying_type_data.iloc[:, [idx, idx+12, idx+24]]
        subset[f"{month} Aggregate"] = subset.mean(axis=1)
        aggregate[f"{month} Aggregate"] = subset.iloc[:, -1]

    return aggregate



# get all rows of prepandemic months (0-36) for each type of bullying
combined_pre_pandemic = pivot_data.loc[:, "sch_cy_bly_sum"].iloc[:,[i for i in range(36)]]
school_pre_pandemic = pivot_data.loc[:, "sch_bly"].iloc[:,[i for i in range(36)]]
cyber_pre_pandemic = pivot_data.loc[:, "cy_bly"].iloc[:,[i for i in range(36)]]

# aggregate
combined_agg = aggregate_pre_pandemic(combined_pre_pandemic)
cyber_agg = aggregate_pre_pandemic(cyber_pre_pandemic)
school_agg = aggregate_pre_pandemic(school_pre_pandemic)
combined_agg
```

```python
def pct_change_calculation(pandemic, pre_pandemic):
    if pre_pandemic == 0:
        return np.nan
    return int(round(100 * ((pandemic - pre_pandemic) / pre_pandemic)))



def get_pct_change_by_state(row, columns, state, pre_pandemic_agg):
    """Calculates percent change with prepandemic data for all cols"""
    
    # map month number to column in aggregate df
    month_mappings = {num+1: col.split()[0] for num, col in enumerate(pre_pandemic_agg.columns)}
    year = 2019
    pct_change_data = pd.DataFrame()
    
    for col in range(len(row)):
        if columns[col].month == 1:
            year += 1
        pct_change = pct_change_calculation(row[col], pre_pandemic_agg.loc[state, f"{month_mappings[columns[col].month]} Aggregate"])
        pct_change_data.loc[state, f"{month_mappings[columns[col].month]} {year}"] = pct_change
    return pct_change_data
```

```python
def get_pct_change_data(bullying_type):
    data = pd.read_csv("bullying_viz_data.csv")
    data = data.fillna(0)

    # convert dates to datetime
    data['date'] = pd.to_datetime(data['date'])

    # pivot by state, so we can compare across time
    pivot_data = data.pivot(index='state', columns = 'date')
    
    # get pre pandemic data
    pre_pandemic = pivot_data.loc[:, bullying_type].iloc[:,[i for i in range(36)]]
    
    # aggregate all years of pre pandemic data
    pre_pandemic_agg = aggregate_pre_pandemic(pre_pandemic)
    
    # get pandemic data
    pandemic = pivot_data.loc[:, bullying_type].iloc[:,[i for i in range(38, len(pivot_data.loc[:, bullying_type].iloc[0]))]]
    
    # calculate pct change for all states
    pct_change_states = []
    for state in pandemic.index:
        state_data = get_pct_change_by_state(pandemic.loc[state], pandemic.columns, state, pre_pandemic_agg)
        pct_change_states.append(state_data)
    return pd.concat(pct_change_states)
    
```

```python
pct_change_states = []
for state in combined_pandemic.index:
    state_data = get_pct_change_by_state(combined_pandemic.loc[state], combined_pandemic.columns, state, combined_agg)
    pct_change_states.append(state_data)
pd.concat(pct_change_states)
```

```python
import pandas as pd
import numpy as np

data = pd.read_csv("bullying_viz_data.csv")
data = data.fillna(0)
    
# convert dates to datetime
data['date'] = pd.to_datetime(data['date'])
data
```

```python
states = data.groupby('state')
states_names = states.groups

all_states = {}
for s in states_names:
    all_states[s] = states.get_group(s)
    
alaska = all_states['AL']
alaska
```

```python
# get baseline values for school bullying and cyberbullying searches in March 2020
baseline_sch_bly = alaska.loc[alaska['date']=='2020-03-01']['sch_bly'].values[0]
baseline_cy_bly = alaska.loc[alaska['date']=='2020-03-01']['cy_bly'].values[0]
```

```python
from percent_change import pct_change_calculation

alaska['pct_change_sch'] = alaska['sch_bly'].apply(lambda x: pct_change_calculation(x, baseline_sch_bly))
alaska['pct_change_cy'] = alaska['cy_bly'].apply(lambda x: pct_change_calculation(x, baseline_cy_bly))
```

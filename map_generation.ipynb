```python
import pandas as pd
import numpy as np
import plotly
import os
from percent_change import reformat_to_pct_change

cwd = os.getcwd()
data = pd.read_csv("bullying_viz_data.csv")
data
```

### Single Plot

```python
percent_change_data = reformat_to_pct_change(data)
percent_change_data
```

```python
scale = [[0.0, "#e6d8d8"],[0.2, '#ff9999'],[0.4, '#ff4d4d'], 
       [0.6, '#ff1a1a'],[0.8, '#cc0000'],[1.0, '#4d0000']] # reds

slider = []
dates = []

# generate plot for each year of data
for col in percent_change_data.columns:
    if col[0] == 'sch_cy_bly_sum':  # change string for different versions of data
        data_plot = dict(
            type='choropleth', 
            visible=False,
            colorscale = scale,
            autocolorscale = False,
            locations = percent_change_data.index, 
            z = percent_change_data[col].astype(float), 
            locationmode = 'USA-states',
            marker = dict(     # for the lines separating states
                        line = dict (
                                  color = 'rgb(255,255,255)', 
                                  width = 2) ),               
            colorbar = dict(
                        title = "Increase")
            ) 
        
        date = f"{col[1]}"
        dates.append(date)
        slider.append(data_plot)
```

```python
# generate slider
steps = []
for i in range(len(slider)):
    for j in range(len(slider)):
        step = dict(method='restyle',
                    args=['visible', [False] * len(slider)],
                    label=dates[i]) 
        step['args'][1][i] = True
        steps.append(step)
        
sliders = [dict(active=0, pad={"t": 1}, steps=steps,)]  
```

```python
layout = dict(geo=dict(scope='usa',
                       projection={'type': 'albers usa'}),
              sliders=sliders,
             title="Percent Change in Sum of Bullying Searches") # have to change this when adding 


fig = dict(data=slider, layout=layout) 

plotly.offline.iplot(fig)
```

```python
plotly.io.write_html(fig, file='sum_bullying_slider.html', auto_open=True)
```
